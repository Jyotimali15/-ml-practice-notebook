```python
import numpy as np
from sklearn.impute import SimpleImputer
matrix = np.array([
    [1.0, 2.0, np.nan],
    [4.0, np.nan, 6.0],
    [np.nan, 8.0, 9.0]
])
imputer = SimpleImputer(strategy='mean')

imputer.fit(matrix)
means = imputer.statistics_


imputed_matrix = imputer.transform(matrix)

print(f"Matrix shape: {matrix.shape}")
print(f"Means: {means}")
print(f"Imputed:\n{imputed_matrix}")
```
